```python
#Nama : Inas Fikri Dzakiyyah
#NIM : 4232511056
#Tanggal : 13-11-26
#Mata Kuliah : Pemrograman Dasar
#Soal 9

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]
data_tidak_valid = 0

for n, teks in enumerate(data, start=1):
    try:
        nilai = float(teks)
    except ValueError:
        print(f"Unit {n}: data tidak valid")
        data_tidak_valid += 1
        continue

    if nilai > 2.0:
        print(f"Warning: Unit {n} vibrasi tinggi!")
        if nilai > 3.0:
            print("KRITIS: pemeriksaan dihentikan")
            break

print(f"Jumlah data tidak valid: {data_tidak_valid}")
```

```python
#Nama : Inas Fikri Dzakiyyah
#NIM : 4232511056
#Tanggal : 13-11-26
#Mata Kuliah : Pemrograman Dasar
#Soal 9

kapasitas = 100.0
daya_panel = ["15", "22", "30", "x", "28", "25"]
isi = 0.0
riwayat = []

for jam, teks in enumerate(daya_panel, start=1):
    try:
        energi = float(teks)
    except ValueError:
        print(f"Jam {jam}: data panel tidak valid, dilewati")
        continue

    isi += energi
    if isi >= kapasitas:
        isi = kapasitas
        riwayat.append(100.0)
        print(f"Jam {jam}: baterai PENUH (100.0%) - pengisian dihentikan")
        break

    persen = isi / kapasitas * 100
    riwayat.append(persen)
    if persen < 40:
        status = "RENDAH"
    elif persen <= 80:
        status = "SEDANG"
    else:
        status = "TINGGI"
    print(f"Jam {jam}: {isi:.1f} kWh ({persen:.1f}%) -> {status}")
```
